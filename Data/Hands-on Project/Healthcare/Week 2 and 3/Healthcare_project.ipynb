Data Preprocessing: Clean and Preprocess the data

```python
# Check for missing values in each dataset
print("Missing values in Patient Demographics:")
print(patient_data.isnull().sum())

print("\nMissing values in Vitals:")
print(vitals_data.isnull().sum())

print("\nMissing values in Health Risk:")
print(Health_risk_data.isnull().sum())
```

```python
#Ensure that reading_time and assessment_time columns are datetime objects

# Convert reading_time to datetime in the vitals DataFrame
vitals_data['reading_time'] = pd.to_datetime(vitals_data['reading_time'])

# Convert assessment_time to datetime in the health_risk DataFrame
Health_risk_data['Assessment_time'] = pd.to_datetime(Health_risk_data['Assessment_time'])

# Check the data types to confirm
print(vitals_data.dtypes)
print(Health_risk_data.dtypes)
```

```python
# Normalize the systolic blood pressure
vitals_data['normalized_bp'] = vitals_data['blood_pressure'].apply(lambda x: int(x.split('/')[0])) / 120

print(vitals_data.head)
```

```python
#Drop any duplicate rows in the data

vitals_data = vitals_data.drop_duplicates()
patient_data = patient_data.drop_duplicates()
Health_risk_data = Health_risk_data.drop_duplicates()
```

```python
# Ensure Risk_score is between 0 and 1
Ensure_Risk_score = Health_risk_data[(Health_risk_data['Risk_score'] >= 0) & (Health_risk_data['Risk_score'] <= 1)]
```

```python
#Display the cleaned data

print(patient_data)
print(vitals_data)
print(Health_risk_data)
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, split

#Create SparkSession for streaming patient vitals data
spark =  SparkSession.builder.appName("PatientMonitoring").getOrCreate()
```

```python
#Read the csv file
vitals_df = spark.read.csv('/content/sample_data/vitals_fact.csv', header=True, inferSchema=True)
```

```python
# Extract systolic and diastolic blood pressure
vitals_df = vitals_df.withColumn("systolic_bp", split(col("blood_pressure"), '/').getItem(0).cast("int"))
vitals_df = vitals_df.withColumn("diastolic_bp", split(col("blood_pressure"), '/').getItem(1).cast("int"))

# Detect anomalies: identify abnormal heart rate or blood pressure levels
anomalies = vitals_df.filter((col("heart_rate") > 100) | (col("heart_rate") < 50))

anomalies.show()
```
