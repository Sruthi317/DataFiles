```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum as sum_, avg, count, max as max_, min as min_, desc

# Create Spark session
spark = SparkSession.builder .appName("EcommerceRealTime") .getOrCreate()

#Read the csv file
order_df = spark.read.csv('/content/sample_data/order_data.csv', header=True, inferSchema=True)

order_df.show()
```

```python
 #Perform real-time analysis using PySpark


#Calculate total sales for each product
total_sales = order_df.groupBy("product_id").agg(sum_("order_amount").alias("total_sales"))

total_sales.show()

# Calculate average order amount for each product
average_order_amount = order_df.groupBy("product_id") \
    .agg(avg("order_amount").alias("average_order_amount"))

average_order_amount.show()

# Calculate order count for each product
order_count = order_df.groupBy("product_id") \
    .agg(count("order_id").alias("order_count"))

order_count.show()

# Calculate total quantity sold for each product
total_quantity_sold = order_df.groupBy("product_id") \
    .agg(sum_("quantity").alias("total_quantity_sold"))

total_quantity_sold.show()

# Calculate maximum order amount for each product
max_order_amount = order_df.groupBy("product_id") \
    .agg(max_("order_amount").alias("max_order_amount"))

max_order_amount.show()

# Calculate minimum order amount for each product
min_order_amount = order_df.groupBy("product_id") \
    .agg(min_("order_amount").alias("min_order_amount"))

min_order_amount.show()

# Calculate unique customer count for each product
unique_customer_count = order_df.groupBy("product_id") \
    .agg(count("customer_id").alias("unique_customer_count"))

unique_customer_count.show()

```
